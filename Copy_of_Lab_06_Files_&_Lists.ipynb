## **Part A** - Creating a file
---

Here we will be retrieving data from the user input, adding it to a list, and then publishing it to a file so that it can persist after the program has been exited.<br/><br/>

### Your Task

Create a basic loop that reads in scores from the user until the user enters the sentinel value of -1. Each score should be added to a list. <br/><br/>

After the loop has been exited, the program should take the top 10 scores and print them in order (highest to lowest) to the screen, and also to a file called `highscores.txt` and then the program should exit.

<br/>


### Your Answer

```python
# Student Name: Ilaf Al Kurdi
# Student Number: 1038115

scores = []

score = int(input("Enter a score (-1 to finish): "))

while score != -1:
    scores.append(score)
    score = int(input("Enter a score (-1 to finish): "))

scores.sort(reverse=True)

top_scores = scores[:10]

print("\nTop 10 Highscores:")

for score in top_scores:
    print(score)

f = open("highscores.txt", "w")

for score in top_scores:
    f.write(str(score) + "\n")

f.close()
```

## **Part B** - Data Permanence  
---

Here we will add the second part of retaining a highscores list - being able to read them back into the program.<br/><br/>

### Your Task

Copy your code from Part A and paste it into the code block below.<br/><br/>
Next, you need to extend the program to:
* open the `highscores.txt` file
* read and print the current highscores list to the screen
* populate the highscores list<br/><br/>

Once this is done, you need to ensure that the functionality in Part A still works - that is, that you can take in scores via the loop and add them to the list, and then re-save them to the file.




<br/>


### Your Answer

```python
# Student Name: Ilaf Al Kurdi
# Student Number: 1038115

scores = []

try:
    f = open("highscores.txt", "r")

    for line in f:
        scores.append(int(line.strip()))

    f.close()

except FileNotFoundError:
    print("No previous highscores found.")


print("Current Highscores:")

for score in scores:
    print(score)


score = int(input("\nEnter a score (-1 to finish): "))

while score != -1:
    scores.append(score)
    score = int(input("Enter a score (-1 to finish): "))


scores.sort(reverse=True)

scores = scores[:10]


print("\nUpdated Top 10 Highscores:")

for score in scores:
    print(score)


f = open("highscores.txt", "w")

for score in scores:
    f.write(str(score) + "\n")

f.close()
```

### Your Answer

```python
# Student Name: Ilaf Al Kurdi
# Student Number: 1038115

from datetime import datetime

scores = []


def load_scores():
    try:
        f = open("highscores.txt", "r")

        for line in f:
            line = line.strip()

            if line != "":
                data = line.split(",")

                if len(data) == 1:
                    score = int(data[0])
                    scores.append(["Unknown", score, "Previous score"])

                elif len(data) >= 3:
                    name = data[0]
                    score = int(data[1])
                    date_time = data[2]

                    scores.append([name, score, date_time])

        f.close()

    except FileNotFoundError:
        print("No previous highscores found.")


def save_scores():
    f = open("highscores.txt", "w")

    for item in scores:
        f.write(
            item[0] + "," +
            str(item[1]) + "," +
            item[2] + "\n"
        )

    f.close()


def sort_scores():
    scores.sort(
        key=lambda item: item[1],
        reverse=True
    )

    del scores[10:]


def view_scores():
    sort_scores()

    print("\nHighscores:")
    print("------------------------------")

    if len(scores) == 0:
        print("No scores available.")

    else:
        for i in range(len(scores)):
            print(
                str(i + 1) + ".",
                scores[i][0],
                "-",
                scores[i][1],
                "-",
                scores[i][2]
            )


def add_score():
    name = input("Enter player name: ")
    score = int(input("Enter score: "))

    date_time = datetime.now().strftime("%d/%m/%Y %H:%M")

    scores.append([name, score, date_time])

    sort_scores()

    print("Score added.")


def delete_score():
    view_scores()

    if len(scores) > 0:
        number = int(input("\nEnter the score number to delete: "))

        if number >= 1 and number <= len(scores):
            scores.pop(number - 1)
            print("Score deleted.")
        else:
            print("Invalid number.")


def modify_score():
    view_scores()

    if len(scores) > 0:
        number = int(input("\nEnter the score number to modify: "))

        if number >= 1 and number <= len(scores):
            new_name = input("Enter new player name: ")
            new_score = int(input("Enter new score: "))

            date_time = datetime.now().strftime("%d/%m/%Y %H:%M")

            scores[number - 1] = [
                new_name,
                new_score,
                date_time
            ]

            sort_scores()

            print("Score updated.")

        else:
            print("Invalid number.")


def clear_scores():
    scores.clear()
    print("Highscore list cleared.")


load_scores()


while True:

    print("\nHighscores Menu")
    print("1. Add new score")
    print("2. View scores")
    print("3. Delete score")
    print("4. Modify score")
    print("5. Clear highscore list")
    print("6. Exit & Save")

    choice = input("Choose an option: ")

    if choice == "1":
        add_score()

    elif choice == "2":
        view_scores()

    elif choice == "3":
        delete_score()

    elif choice == "4":
        modify_score()

    elif choice == "5":
        clear_scores()

    elif choice == "6":
        save_scores()
        print("Highscores saved.")
        break

    else:
        print("Invalid option.")
```
